```python
print('=====Welcome to Basic calculator (+,-,*,/)=====')
def add(x,y):
    return x+y
    print(x+y)

def subtract(x,y):
    return x-y
    print(x-y)


def divide(x,y):
    return x/y
    print(x/y)


def multiply(x,y):
    return x*y
    print(x*y)
    
```

```python
choice=input('Enter choice; add, subtract, divide or multiply')
num1=float(input('Enter the first number'))
num2=float(input('Enter second number'))
```

```python
while True:
    choice=input('Enter choice; add, subtract, divide or multiply')
    
    if choice in('add','subtract','divide','multiply'):
        num1=float(input('Enter the first number'))
        num2=float(input('Enter second number'))
    
                 
        if choice=='add':
            print(num1, ' + ', num2,' =',add(num1,num2))
        
        elif choice=='subtract':
            print(num1, ' - ', num2,' =',subtract(num1,num2))
        
        elif choice=='divide':
            print(num1, ' / ', num2,' =',divide(num1,num2))
        
        elif choice=='multiply':
            print(num1, ' * ', num2,' =',multiply(num1,num2))
    
    else:
        print('Wrong input, kindly choose from the four options')
        continue


    next_calculation=input('Want to do another calculation? Y/N')
    if next_calculation=='y' or next_calculation=='Y':
        continue
    else:
        print('===Goodbye===')
        break
    
        
   
```
